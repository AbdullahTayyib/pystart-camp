### Soru 1: Kargo Ücreti

Kullanıcıdan paketin ağırlığını (kg), gönderim mesafesini (km) ve hızlı kargo isteyip istemediğini (`e/h`) al. Ücreti şu tarifeye göre hesapla:

| Ağırlık | Ücret |
| :--- | :--- |
| 2 kg'a kadar (2 dahil) | 50 TL |
| 2-5 kg (5 dahil) | 80 TL |
| 5-10 kg (10 dahil) | 120 TL |
| 10 kg üstü | 120 TL + 10 kg'ı aşan her kg için 10 TL |

- Mesafe 500 km'den fazlaysa ücrete **%20** ekle.
- Hızlı kargo seçildiyse ücreti **1.5** ile çarp.
- Ağırlık ya da mesafe 0 veya negatifse ücret hesaplama, `'Geçersiz ağırlık ya da mesafe.'` yaz.

**Örnek Çıktı:**
```text
Paketin ağırlığı (kg): 12.5
Mesafe (km): 600
Hızlı kargo olsun mu? (e/h): e
Kargo ücreti: 261.00 TL
```

> **Kontrol:** Kodunu sınır değerlerle de dene: 2 kg 50 TL, 2.1 kg 80 TL tutmalı.

````python
agirlik = float(input("Paketin ağırlığı (kg): "))
mesafe = float(input("Mesafe (km): "))
hizli_kargo = input("Hızlı kargo olsun mu? (e/h): ").strip().lower()

if agirlik <= 0 or mesafe <= 0:
    print("Geçersiz ağırlık ya da mesafe.")
else:

    if agirlik <= 2:
        ucret = 50.0
    elif agirlik <= 5:
        ucret = 80.0
    elif agirlik <= 10:
        ucret = 120.0
    else:
        ucret = 120.0 + (agirlik - 10) * 10

    if mesafe > 500:
        ucret += ucret * 0.20

    if hizli_kargo == "e":
        ucret *= 1.5

    print(f"Kargo ücreti: {ucret:.2f} TL")
````

### Soru 2: Döngüler

İki kısım, tek hücrede:

1. Kullanıcıdan bir sayı al ve 1'den 10'a kadar çarpım tablosunu `for` ve `range()` ile yazdır.
2. Kullanıcı `'q'` yazana kadar `while` ile sayı iste. Sayı olmayan bir girişte `'Geçersiz giriş'` yazıp devam et (`continue`), `'q'` gelince döngüden çık (`break`). Sonunda girilen sayıların **toplamını** ve **ortalamasını** yazdır.

**Örnek Çıktı (2. kısım):**
```text
Sayı gir (bitirmek için q): 3
```

````python
sayi = int(input("Çarpım tablosu için bir sayı giriniz: "))
print(f"\n--- {sayi} Sayısının Çarpım Tablosu ---")

for i in range(1, 11):
    print(f"{sayi} x {i} = {sayi * i}")

print("\n" + "=" * 30 + "\n")

toplam = 0
adet = 0

while True:
    girdi = input("Sayı gir (bitirmek için q): ").strip()

    if girdi.lower() == "q":
        break

    try:
        girilen_sayi = float(girdi)
        toplam += girilen_sayi
        adet += 1
    except ValueError:
        print("Geçersiz giriş, tekrar dene.")
        continue

if adet > 0:
    ortalama = toplam / adet
    print(f"Toplam: {toplam:g} | Ortalama: {ortalama:.2f}")
else:
    print("Hiç geçerli sayı girilmedi.")
````

### Soru 3: List Comprehension, zip ve enumerate

1. 1 ile 30 arasındaki çift sayıların karelerinden oluşan listeyi tek satırlık bir `list comprehension` ile oluştur ve yazdır.
2. `sicakliklar = [12, 15, 9, 18, 21, 17, 14]` listesindeki Celsius değerlerini list comprehension ile Fahrenheit'a çevir ($F = C \times 9/5 + 32$) ve bir basamağa yuvarla.
3. `isimler = ['Ali', 'Ayşe', 'Can']` ve `yaslar = [20, 19, 21]` listelerini `zip()` ile eşleştir, `enumerate()` ile numaralandırarak yazdır.

**Örnek Çıktı (2. ve 3. kısım):**
```text
[53.6, 59.0, 48.2, 64.4, 69.8, 62.6, 57.2]
1. Ali - 20 yaşında
2. Ayşe - 19 yaşında
3. Can - 21 yaşında
```

````python
cift_kareler = [x**2 for x in range(1, 30) if x % 2 == 0]
print("1-30 Arası Çift Sayıların Kareleri:")
print(cift_kareler)
print()

sicakliklar = [12, 15, 9, 18, 21, 17, 14]
fahrenheit = [round(c * 9 / 5 + 32, 1) for c in sicakliklar]
print("Fahrenheit Değerleri:")
print(fahrenheit)
print()

isimler = ["Ali", "Ayşe", "Can"]
yaslar = [20, 19, 21]

print("Kişi Listesi:")
for sira, (isim, yas) in enumerate(zip(isimler, yaslar), start=1):
    print(f"{sira}. {isim} - {yas} yaşında")

````

### Soru 4: Fonksiyonlar

Aşağıdaki fonksiyonları yaz ve her birini en az bir kez çağırıp sonucunu yazdır. Fonksiyonlar sonucu ekrana yazmasın, `return` ile döndürsün.

1. `sesli_harf_say(metin)`: Metindeki sesli harf sayısını döndürür. `sesli_harf_say('Python öğreniyorum')` sonucu `6` olmalı.
2. `en_uzun_kelime(cumle)`: Cümledeki en uzun kelimeyi döndürür; eşitlikte ilk geleni seçer. `'Kod yazmak pratik ister'` için sonuç `'yazmak'`.
3. `istatistik(*sayilar)`: İstediğin kadar sayı alır; en küçük, en büyük ve ortalamayı birlikte döndürür. `istatistik(4, 8, 15, 16, 23, 42)` sonucu `(4, 42, 18.0)`.
4. `selamla(isim, dil='tr')`: Dil 'tr' ise 'Merhaba, Ayşe!', 'en' ise 'Hello, Ayşe!' döndürür. Dil verilmezse Türkçe selamlar.
5. Fonksiyon yazmadan: `['kod', 'python', 'döngü', 'if', 'fonksiyon', 'liste']` listesinden `filter()` ve `lambda` ile 4 harften uzun kelimeleri seç, `map()` ve `lambda` ile büyük harfe çevir.

````python
def sesli_harf_say(metin):
    sesliler = "aeıioöuüAEIİOÖUÜ"
    sayac = 0
    for harf in metin:
        if harf in sesliler:
            sayac += 1
    return sayac

def en_uzun_kelime(cumle):
    kelimeler = cumle.split()
    if not kelimeler:
        return ""

    en_uzun = kelimeler[0]
    for kelime in kelimeler:
        if len(kelime) > len(en_uzun):
            en_uzun = kelime
    return en_uzun

def istatistik(*sayilar):
    if not sayilar:
        return None
    en_kucuk = min(sayilar)
    en_buyuk = max(sayilar)
    ortalama = sum(sayilar) / len(sayilar)
    return (en_kucuk, en_buyuk, ortalama)

def selamla(isim, dil="tr"):
    if dil == "tr":
        return f"Merhaba, {isim}!"
    elif dil == "en":
        return f"Hello, {isim}!"
    else:
        return f"Merhaba, {isim}!"

print("1. Sesli Harf Sayısı:", sesli_harf_say("Python öğreniyorum"))
print("2. En Uzun Kelime:", en_uzun_kelime("Kod yazmak pratik ister"))
print("3. İstatistikler:", istatistik(4, 8, 15, 16, 23, 42))
print("4. Selamlama (TR):", selamla("Ayşe"))
print("4. Selamlama (EN):", selamla("Ayşe", dil="en"))

kelime_listesi = ["kod", "python", "döngü", "if", "fonksiyon", "liste"]

uzun_kelimeler = filter(lambda k: len(k) > 4, kelime_listesi)

buyuk_kelimeler = list(map(lambda k: k.upper(), uzun_kelimeler))

print("5. Filtrelenmiş ve Büyütülmüş Kelimeler:", buyuk_kelimeler)

````
