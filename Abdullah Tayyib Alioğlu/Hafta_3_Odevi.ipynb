### Soru 1: Kitap Sınıfı

Aşağıdaki özellik ve metotlara sahip `Kitap` sınıfını yaz:

- **Özellikler:** `baslik`, `yazar`, `yil` ve `odunc_mu` (başta `False`).
- `odunc_ver()`: Kitabı ödünçte olarak işaretler.
- `iade_et()`: Kitabı rafta olarak işaretler.
- `__str__()`: Kitabı tek satırda gösterir: `Suç ve Ceza - Fyodor Dostoyevski (1866) [Rafta]`

> İki kitap oluştur, birini ödünç ver ve ikisini de `print()` ile yazdırarak sınıfını dene.

````python
# --- Soru 1 Kod Alanı ---
class Kitap:
    def __init__(self, baslik, yazar, yil):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = False

    def odunc_ver(self):
        self.odunc_mu = True

    def iade_et(self):
        self.odunc_mu = False

    def __str__(self):
        durum = "Ödünçte" if self.odunc_mu else "Rafta"
        return f"{self.baslik} - {self.yazar} ({self.yil}) [{durum}]"

kitap1 = Kitap("Suç ve Ceza", "Fyodor Dostoyevski", 1866)
kitap2 = Kitap("Simyacı", "Paulo Coelho", 1988)

kitap1.odunc_ver()

print(kitap1)
print(kitap2)
pass

# Test:
````

### Soru 2: Kutuphane Sınıfı ve Temel Menü

`Kutuphane` sınıfını ve menüyü tasarla:

- Kitapları bir `kitaplar` listesinde tutar.
- `kitap_ekle(kitap)`: Listeye kitap ekler.
- `kitap_bul(baslik)`: Başlığa göre kitabı bulur (büyük-küçük harf fark etmez).
- `ara(kelime)`: Başlığında kelime geçen kitapların listesini döndürür.
- `odunc_ver(baslik)` ve `iade_al(baslik)`: Kitabı bulup ilgili `Kitap` metodunu çağırır.

Ayrıca ekleme, listeleme, ödünç verme, iade alma, arama ve çıkış seçenekleri olan bir menü yaz. Menü, çıkış seçilene kadar tekrar gösterilsin.

````python
# --- Soru 2 Kod Alanı ---
class Kutuphane:
    def __init__(self):
        self.kitaplar = []

    def kitap_ekle(self, kitap):
        self.kitaplar.append(kitap)

    def kitap_bul(self, baslik):
        for kitap in self.kitaplar:
            if kitap.baslik.lower() == baslik.lower():
                return kitap
        return None

    def ara(self, kelime):
        sonuclar = []
        for kitap in self.kitaplar:
            if kelime.lower() in kitap.baslik.lower():
                sonuclar.append(kitap)
        return sonuclar

    def odunc_ver(self, baslik):
        kitap = self.kitap_bul(baslik)
        if kitap:
            kitap.odunc_ver()

    def iade_al(self, baslik):
        kitap = self.kitap_bul(baslik)
        if kitap:
            kitap.iade_et()

kutuphane = Kutuphane()
kutuphane.kitap_ekle(Kitap("Suç ve Ceza", "Fyodor Dostoyevski", 1866))
kutuphane.kitap_ekle(Kitap("Simyacı", "Paulo Coelho", 1988))
kutuphane.kitap_ekle(Kitap("Kürk Mantolu Madonna", "Sabahattin Ali", 1943))

while True:
    print("\n1) Kitap Ekle")
    print("2) Listele")
    print("3) Ödünç Ver")
    print("4) İade Al")
    print("5) Ara")
    print("0) Çıkış")
    
    secim = input("Seçimin: ")
    
    if secim == "1":
        baslik = input("Kitap başlığı: ")
        yazar = input("Yazar: ")
        yil = int(input("Yıl: "))
        yeni_kitap = Kitap(baslik, yazar, yil)
        kutuphane.kitap_ekle(yeni_kitap)
        print("Kitap eklendi.")
    elif secim == "2":
        for kitap in kutuphane.kitaplar:
            print(kitap)
    elif secim == "3":
        baslik = input("Ödünç verilecek kitap: ")
        kutuphane.odunc_ver(baslik)
        print("İşlem tamamlandı.")
    elif secim == "4":
        baslik = input("İade edilecek kitap: ")
        kutuphane.iade_al(baslik)
        print("İşlem tamamlandı.")
    elif secim == "5":
        kelime = input("Aranan kelime: ")
        sonuclar = kutuphane.ara(kelime)
        for kitap in sonuclar:
            print(kitap)
    elif secim == "0":
        break
    pass

# Menü:
````

### Soru 3: Hata Yönetimi

Özel hata sınıfları tanımla ve işlemleri güvenli hale getir. Soru 1 ve 2'deki metotları buna göre güncelle.

- Kendi hata sınıflarını yaz: `KitapBulunamadi` ve `KitapMusaitDegil` (`Exception` sınıfından türesin).
- `odunc_ver()` kitap zaten ödünçteyse, `iade_et()` kitap zaten raftaysa `KitapMusaitDegil` fırlatsın (`raise`).
- `kitap_bul()` kitabı bulamazsa `KitapBulunamadi` fırlatsın.
- Menüde bu hataları `try / except` ile yakala ve anlaşılır mesaj göster. Program çökmeksizin devam etsin.
- Yıl sayı olarak girilmezse (`ValueError`) uyarı ver.

````python
# --- Soru 3 Kod Alanı ---
class KitapBulunamadi(Exception):
    pass

class KitapMusaitDegil(Exception):
    pass

class Kitap:
    def __init__(self, baslik, yazar, yil):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = False

    def odunc_ver(self):
        if self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten ödünç verilmiş.")
        self.odunc_mu = True

    def iade_et(self):
        if not self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten rafta.")
        self.odunc_mu = False

    def __str__(self):
        durum = "Ödünçte" if self.odunc_mu else "Rafta"
        return f"{self.baslik} - {self.yazar} ({self.yil}) [{durum}]"

class Kutuphane:
    def __init__(self):
        self.kitaplar = []

    def kitap_ekle(self, kitap):
        self.kitaplar.append(kitap)

    def kitap_bul(self, baslik):
        for kitap in self.kitaplar:
            if kitap.baslik.lower() == baslik.lower():
                return kitap
        raise KitapBulunamadi(f"'{baslik}' isimli kitap bulunamadı.")

    def ara(self, kelime):
        sonuclar = []
        for kitap in self.kitaplar:
            if kelime.lower() in kitap.baslik.lower():
                sonuclar.append(kitap)
        return sonuclar

    def odunc_ver(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.odunc_ver()

    def iade_al(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.iade_et()

kutuphane = Kutuphane()
kutuphane.kitap_ekle(Kitap("Suç ve Ceza", "Fyodor Dostoyevski", 1866))
kutuphane.kitap_ekle(Kitap("Simyacı", "Paulo Coelho", 1988))
kutuphane.kitap_ekle(Kitap("Kürk Mantolu Madonna", "Sabahattin Ali", 1943))

while True:
    print("\n1) Kitap Ekle")
    print("2) Listele")
    print("3) Ödünç Ver")
    print("4) İade Al")
    print("5) Ara")
    print("0) Çıkış")
    
    secim = input("Seçimin: ")
    
    try:
        if secim == "1":
            baslik = input("Kitap başlığı: ")
            yazar = input("Yazar: ")
            yil_input = input("Yıl: ")
            if not yil_input.isdigit():
                raise ValueError("Yıl sayısal bir değer olmalıdır.")
            yil = int(yil_input)
            yeni_kitap = Kitap(baslik, yazar, yil)
            kutuphane.kitap_ekle(yeni_kitap)
            print("Kitap eklendi.")
        elif secim == "2":
            for kitap in kutuphane.kitaplar:
                print(kitap)
        elif secim == "3":
            baslik = input("Ödünç verilecek kitap: ")
            kutuphane.odunc_ver(baslik)
            print("Ödünç verildi.")
        elif secim == "4":
            baslik = input("İade edilecek kitap: ")
            kutuphane.iade_al(baslik)
            print("İade alındı.")
        elif secim == "5":
            kelime = input("Aranan kelime: ")
            sonuclar = kutuphane.ara(kelime)
            for kitap in sonuclar:
                print(kitap)
        elif secim == "0":
            break
    except (KitapBulunamadi, KitapMusaitDegil, ValueError) as hata:
        print(f"Hata: {hata}") 
````

### Soru 4: Dosya İşlemleri (kitaplar.txt)

Kitapları kalıcı olarak dosyada sakla:

- Her kitap `kitaplar.txt` dosyasında bir satır olur. Alanlar `;` ile ayrılır, aralarında boşluk yoktur; son alan kitap ödünçteyse `1`, raftaysa `0` olur:
  ```text
  Suç ve Ceza;Fyodor Dostoyevski;1866;0
  Simyacı;Paulo Coelho;1988;1
  Kürk Mantolu Madonna;Sabahattin Ali;1943;0
  ```
- `Kutuphane` sınıfına kitapları dosyaya yazan `kaydet()` ve dosyadan okuyan `yukle()` metotlarını ekle.
- Program açılınca dosyadaki kitapları yükler. Dosya yoksa (`FileNotFoundError`) boş başlar.
- Menüdeki çıkış seçeneği `'Kaydet ve çık'` olsun ve kitapları dosyaya yazsın (`encoding='utf-8'`).
- Teslim etmeden önce dosyada en az 3 kitap olsun.

````python
# --- Soru 4 Kod Alanı ---
class KitapBulunamadi(Exception):
    pass

class KitapMusaitDegil(Exception):
    pass

class Kitap:
    def __init__(self, baslik, yazar, yil, odunc_mu=False):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = odunc_mu

    def odunc_ver(self):
        if self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten ödünç verilmiş.")
        self.odunc_mu = True

    def iade_et(self):
        if not self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten rafta.")
        self.odunc_mu = False

    def __str__(self):
        durum = "Ödünçte" if self.odunc_mu else "Rafta"
        return f"{self.baslik} - {self.yazar} ({self.yil}) [{durum}]"

class Kutuphane:
    def __init__(self):
        self.kitaplar = []
        self.yukle()

    def kitap_ekle(self, kitap):
        self.kitaplar.append(kitap)

    def kitap_bul(self, baslik):
        for kitap in self.kitaplar:
            if kitap.baslik.lower() == baslik.lower():
                return kitap
        raise KitapBulunamadi(f"'{baslik}' isimli kitap bulunamadı.")

    def ara(self, kelime):
        sonuclar = []
        for kitap in self.kitaplar:
            if kelime.lower() in kitap.baslik.lower():
                sonuclar.append(kitap)
        return sonuclar

    def odunc_ver(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.odunc_ver()

    def iade_al(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.iade_et()

    def kaydet(self):
        with open("kitaplar.txt", "w", encoding="utf-8") as dosya:
            for kitap in self.kitaplar:
                durum = "1" if kitap.odunc_mu else "0"
                dosya.write(f"{kitap.baslik};{kitap.yazar};{kitap.yil};{durum}\n")

    def yukle(self):
        try:
            with open("kitaplar.txt", "r", encoding="utf-8") as dosya:
                satirlar = dosya.readlines()
                for satir in satirlar:
                    parcalar = satir.strip().split(";")
                    if len(parcalar) == 4:
                        baslik, yazar, yil, durum = parcalar
                        odunc_mu = True if durum == "1" else False
                        kitap = Kitap(baslik, yazar, int(yil), odunc_mu)
                        self.kitaplar.append(kitap)
            print(f"{len(self.kitaplar)} kitap yüklendi.")
        except FileNotFoundError:
            print("Kayıt dosyası bulunamadı, yeni liste oluşturuldu.")

kutuphane = Kutuphane()
if len(kutuphane.kitaplar) == 0:
    kutuphane.kitap_ekle(Kitap("Suç ve Ceza", "Fyodor Dostoyevski", 1866))
    kutuphane.kitap_ekle(Kitap("Simyacı", "Paulo Coelho", 1988))
    kutuphane.kitap_ekle(Kitap("Kürk Mantolu Madonna", "Sabahattin Ali", 1943))
    kutuphane.kaydet()

while True:
    print("\n1) Kitap Ekle")
    print("2) Listele")
    print("3) Ödünç Ver")
    print("4) İade Al")
    print("5) Ara")
    print("0) Kaydet ve çık")
    
    secim = input("Seçimin: ")
    
    try:
        if secim == "1":
            baslik = input("Kitap başlığı: ")
            yazar = input("Yazar: ")
            yil_input = input("Yıl: ")
            if not yil_input.isdigit():
                raise ValueError("Yıl sayısal bir değer olmalıdır.")
            yil = int(yil_input)
            yeni_kitap = Kitap(baslik, yazar, yil)
            kutuphane.kitap_ekle(yeni_kitap)
            print("Kitap eklendi.")
        elif secim == "2":
            for kitap in kutuphane.kitaplar:
                print(kitap)
        elif secim == "3":
            baslik = input("Ödünç verilecek kitap: ")
            kutuphane.odunc_ver(baslik)
            print("Ödünç verildi.")
        elif secim == "4":
            baslik = input("İade edilecek kitap: ")
            kutuphane.iade_al(baslik)
            print("İade alındı.")
        elif secim == "5":
            kelime = input("Aranan kelime: ")
            sonuclar = kutuphane.ara(kelime)
            for kitap in sonuclar:
                print(kitap)
        elif secim == "0":
            kutuphane.kaydet()
            print("Kaydedildi. Görüşmek üzere!")
            break
    except (KitapBulunamadi, KitapMusaitDegil, ValueError) as hata:
        print(f"Hata: {hata}")

````

### Soru 5: Iterator, Generator ve Kendi Modülün

Projenin son halini toparla:

- `Kutuphane` sınıfına `__iter__` metodunu ekle; böylece `for kitap in kutuphane:` yazılabilsin. Listelemeyi bu şekilde yap.
- Raftaki kitapları tek tek `yield` eden bir `musait_kitaplar()` generator metodu yaz.
- Menüye raftaki kitapları listeleyen yeni bir seçenek (`6) Raftakiler`) ekle.
- **Kendi modülün:** Hata sınıflarının, `Kitap` ve `Kutuphane` sınıflarının son halini aşağıdaki `%%writefile kutuphane.py` hücresine yaz. Hücre çalışınca kod `kutuphane.py` dosyasına kaydedilir. Menü hücresi sınıfları `from kutuphane import ...` ile alsın.

**Menü Seçenekleri:**
1) Kitap ekle | 2) Listele | 3) Ödünç ver | 4) İade al | 5) Ara | 6) Raftakiler | 0) Kaydet ve çık

**Örnek Çıktı (menü satırları kısaltıldı):**
```text
3 kitap yüklendi.
Seçimin: 3
Ödünç verilecek kitap: Suç ve Ceza
Ödünç verildi.
Seçimin: 3
Ödünç verilecek kitap: Suç ve Ceza
Hata: 'Suç ve Ceza' zaten ödünç verilmiş.
Seçimin: 6
Kürk Mantolu Madonna - Sabahattin Ali (1943) [Rafta]
Seçimin: 0
Kaydedildi. Görüşmek üzere!
```

> Decorator konusunu (12. modül) 3. oturumda bu proje üzerinde birlikte yazacağız.

````python
%%writefile kutuphane.py
class KitapBulunamadi(Exception):
    pass

class KitapMusaitDegil(Exception):
    pass

class Kitap:
    def __init__(self, baslik, yazar, yil, odunc_mu=False):
        self.baslik = baslik
        self.yazar = yazar
        self.yil = yil
        self.odunc_mu = odunc_mu

    def odunc_ver(self):
        if self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten ödünç verilmiş.")
        self.odunc_mu = True

    def iade_et(self):
        if not self.odunc_mu:
            raise KitapMusaitDegil(f"'{self.baslik}' zaten rafta.")
        self.odunc_mu = False

    def __str__(self):
        durum = "Ödünçte" if self.odunc_mu else "Rafta"
        return f"{self.baslik} - {self.yazar} ({self.yil}) [{durum}]"

class Kutuphane:
    def __init__(self):
        self.kitaplar = []
        self.yukle()

    def __iter__(self):
        return iter(self.kitaplar)

    def musait_kitaplar(self):
        for kitap in self.kitaplar:
            if not kitap.odunc_mu:
                yield kitap

    def kitap_ekle(self, kitap):
        self.kitaplar.append(kitap)

    def kitap_bul(self, baslik):
        for kitap in self.kitaplar:
            if kitap.baslik.lower() == baslik.lower():
                return kitap
        raise KitapBulunamadi(f"'{baslik}' isimli kitap bulunamadı.")

    def ara(self, kelime):
        sonuclar = []
        for kitap in self.kitaplar:
            if kelime.lower() in kitap.baslik.lower():
                sonuclar.append(kitap)
        return sonuclar

    def odunc_ver(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.odunc_ver()

    def iade_al(self, baslik):
        kitap = self.kitap_bul(baslik)
        kitap.iade_et()

    def kaydet(self):
        with open("kitaplar.txt", "w", encoding="utf-8") as dosya:
            for kitap in self.kitaplar:
                durum = "1" if kitap.odunc_mu else "0"
                dosya.write(f"{kitap.baslik};{kitap.yazar};{kitap.yil};{durum}\n")

    def yukle(self):
        try:
            with open("kitaplar.txt", "r", encoding="utf-8") as dosya:
                satirlar = dosya.readlines()
                for satir in satirlar:
                    parcalar = satir.strip().split(";")
                    if len(parcalar) == 4:
                        baslik, yazar, yil, durum = parcalar
                        odunc_mu = True if durum == "1" else False
                        kitap = Kitap(baslik, yazar, int(yil), odunc_mu)
                        self.kitaplar.append(kitap)
            print(f"{len(self.kitaplar)} kitap yüklendi.")
        except FileNotFoundError:
            print("Kayıt dosyası bulunamadı, yeni liste oluşturuldu.")

````

````python
# --- Soru 5: Menü ---
%load_ext autoreload
%autoreload 2

from kutuphane import Kutuphane, Kitap, KitapBulunamadi, KitapMusaitDegil

kutuphane = Kutuphane()
if len(kutuphane.kitaplar) == 0:
    kutuphane.kitap_ekle(Kitap("Suç ve Ceza", "Fyodor Dostoyevski", 1866))
    kutuphane.kitap_ekle(Kitap("Simyacı", "Paulo Coelho", 1988))
    kutuphane.kitap_ekle(Kitap("Kürk Mantolu Madonna", "Sabahattin Ali", 1943))
    kutuphane.kaydet()

while True:
    print("\n1) Kitap ekle")
    print("2) Listele")
    print("3) Ödünç ver")
    print("4) İade al")
    print("5) Ara")
    print("6) Raftakiler")
    print("0) Kaydet ve çık")
    
    secim = input("Seçimin: ")
    
    try:
        if secim == "1":
            baslik = input("Kitap başlığı: ")
            yazar = input("Yazar: ")
            yil_input = input("Yıl: ")
            if not yil_input.isdigit():
                raise ValueError("Yıl sayısal bir değer olmalıdır.")
            yil = int(yil_input)
            yeni_kitap = Kitap(baslik, yazar, yil)
            kutuphane.kitap_ekle(yeni_kitap)
            print("Kitap eklendi.")
        elif secim == "2":
            for kitap in kutuphane:
                print(kitap)
        elif secim == "3":
            baslik = input("Ödünç verilecek kitap: ")
            kutuphane.odunc_ver(baslik)
            print("Ödünç verildi.")
        elif secim == "4":
            baslik = input("İade edilecek kitap: ")
            kutuphane.iade_al(baslik)
            print("İade alındı.")
        elif secim == "5":
            kelime = input("Aranan kelime: ")
            sonuclar = kutuphane.ara(kelime)
            for kitap in sonuclar:
                print(kitap)
        elif secim == "6":
            for kitap in kutuphane.musait_kitaplar():
                print(kitap)
        elif secim == "0":
            kutuphane.kaydet()
            print("Kaydedildi. Görüşmek üzere!")
            break
    except (KitapBulunamadi, KitapMusaitDegil, ValueError) as hata:
        print(f"Hata: {hata}")

````
